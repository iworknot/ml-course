```python
X = df[['vehicle_weight', 'model_year']].head(7)
XT = X.transpose()
XTX = np.matmul(XT, X)
XTX
```

```python
INV = np.linalg.inv(XTX)
INV
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = np.matmul(INV, XT).dot(y)
w
```
